```python
from src.news_analyzer import NewsAnalyzer
from src.technical_analyzer import TechnicalAnalyzer
```

```python
print("📊 TASK 2: Technical Analysis")

# Load data and mock price data
analyzer = NewsAnalyzer('../data/raw_analyst_ratings.csv')

# Create mock price data (replace with real yfinance data)
dates = pd.date_range(start=analyzer.df['date'].min(), end=analyzer.df['date'].max(), freq='D')
mock_prices = 100 + np.cumsum(np.random.normal(0, 1, len(dates)))
price_data = pd.Series(mock_prices, index=dates)
```

```python
# Calculate technical indicators
print("Calculating technical indicators...")

# RSI
rsi = TechnicalAnalyzer.calculate_rsi(price_data, window=14)
print(f"Latest RSI: {rsi.dropna().iloc[-1]:.2f}")

# SMA
sma_20 = TechnicalAnalyzer.calculate_sma(price_data, window=20)
sma_50 = TechnicalAnalyzer.calculate_sma(price_data, window=50)
```

```python
#  Visualize indicators
plt.figure(figsize=(15, 10))

# Price and SMAs
plt.subplot(2, 1, 1)
plt.plot(price_data.index, price_data.values, label='Price', alpha=0.7)
plt.plot(sma_20.index, sma_20.values, label='SMA 20', alpha=0.8)
plt.plot(sma_50.index, sma_50.values, label='SMA 50', alpha=0.8)
plt.title('Price and Moving Averages')
plt.legend()

# RSI
plt.subplot(2, 1, 2)
plt.plot(rsi.index, rsi.values, label='RSI', color='orange')
plt.axhline(y=70, color='r', linestyle='--', alpha=0.7, label='Overbought')
plt.axhline(y=30, color='g', linestyle='--', alpha=0.7, label='Oversold')
plt.title('Relative Strength Index (RSI)')
plt.legend()

plt.tight_layout()
plt.show()
```

```python
#  Save technical results
tech_results = {
    'latest_rsi': rsi.dropna().iloc[-1],
    'latest_sma_20': sma_20.dropna().iloc[-1],
    'latest_sma_50': sma_50.dropna().iloc[-1]
}
pd.DataFrame([tech_results]).to_csv('../results/task2_technical_results.csv', index=False)
print("✅ Task 2 completed - Technical analysis saved")
```
