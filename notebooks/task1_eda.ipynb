```python
import sys
import os
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

```python
sys.path.append(os.path.join(os.getcwd(), '..'))  # Go up one level from notebooks
sys.path.append(os.path.join(os.getcwd(), '../src'))  # Direct to src
from src.news_analyzer import NewsAnalyzer
```

```python
print("🔍 TASK 1: Exploratory Data Analysis")
# Load Data
analyzer = NewsAnalyzer('../data/raw_analyst_ratings.csv')
print(f"Data loaded: {len(analyzer.df)} records")
```

```python
# Basic EDA
analyzer.df['date'] = pd.to_datetime(analyzer.df['date'])
eda_results = analyzer.perform_eda()
print("EDA Results:", eda_results)
```

```python
# Visualizations
plt.figure(figsize=(15, 10))

# Headline length distribution
plt.subplot(2, 2, 1)
analyzer.df['headline_length'].hist(bins=50, alpha=0.7)
plt.title('Headline Length Distribution')

# Top publishers
plt.subplot(2, 2, 2)
analyzer.df['publisher'].value_counts().head(10).plot(kind='barh')
plt.title('Top 10 Publishers')

# Daily article count
plt.subplot(2, 2, 3)
daily_count = analyzer.df.resample('D', on='date').size()
daily_count.plot(title='Daily Article Volume')

plt.tight_layout()
plt.show()
```

```python
#  Save EDA results
eda_results_df = pd.DataFrame([eda_results])
eda_results_df.to_csv('../results/task1_eda_results.csv', index=False)
print("✅ Task 1 completed - EDA results saved")
```
